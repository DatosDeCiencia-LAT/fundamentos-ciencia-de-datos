## 3. Control de flujo (if-else)

Hay ocasiones en las que queremos que nuestro código guarde ciertos valores o realice ciertas operaciones dependiendo de alguna condición. Por ejemplo, supongamos que tenemos un código que queremos que nos indique si una olla de agua que dejamos calentando está hirviendo o no dependiendo de su tempertura asumiendo que su temperatura de ebullición es a los $100^\circ\text{C}$. Es decir, si la olla es tiene un valor igual o mayor a 100 grados celsius, entonces estará ebullendo.

Vamos  medir la tempertura de la olla en dos ocasiones: a los cinco minutos y a los diez minutos de estar calentándose en la estufa. En la **ocasión 1** medimos $67^\circ\text{C}$ y en la **ocasión 2** medimos $100^\circ C$.

La lógica para escribir un control de flujo if-else es la siguiente

```
if condición_que_se_debe_cumplir:
    sucede A
else:
    sucede b
```

````python
ocasion_1 = 67 #grados celsius

ocasion_2 = 100 #grados celsius
````

````python
if ocasion_1 >= 100:
    print("ebullendo!")
else:
    print("no ebullendo.")
````

````python
if ocasion_2 >= 100:
    print("ebullendo!")
else:
    print("no ebullendo.")
````

Ahí es cuando entra en escena el `elif`, que te permite agregar otra condición al control de flujo para tenerla en cuenta en la evaluación. Así, entonces agreguemos una ocasión 3 en la cual dejamos nuestra olla en el congelador por una hora y luego medimos su temperatura, la cuál era de $-4 ^\circ C$, es decir, está congelada. Debemos hacer un código que sea capaz de decir si el agua está ebullendo, temperatura normal, o congelada para los tres casos planteados.

Debe seguir la siguiente lógica

```
if condición_1_sucede:
    haz A
elif condición_2_sucede:
    haz B
else: 
    haz C

````python
ocasion_3 = -4 #grados celcius
````

````python
if ocasion_1 >= 100:
    print("ebullendo!")
elif ocasion_1 < 0:
    print("congelado 🥶")
else:
    print("temperatura normal.")
````

## 4. Loops

### 4.1 For

Hay ocasiones en las que tenemos que realizar una operación repetitivamente. Por ejemplo, supongamos que queremos sumar todos los números dentro de una lista de enteros. Una opción es hacerlo a mano.

````python
lista_sumar = [1,2,3]
````

````python
lista_sumar[0]+ lista_sumar[1] + lista_sumar[2]
````

Sin embargo, cuando nuestra lista tiene más de cien números, esto se convierte en algo muy tedioso y que ocupa mucho tiempo. Es para esto que en los lenguajes de programación hay dos tipos de loop implementados: `for` y `while`.

Un `for` loop hace un recorrido iterativo por los elementos de una lista (o cualquier objeto que contenga multiples elementos) y finaliza cuando los recorre todos. Su lógica es

```
for elemento in lista:
    haz algo con cada elemento
```

Por ejemplo, imprimamos todos los elementos de la lista de enteros.

````python
for el in lista_enteros:
    print(el)
````

Ahora sumemos todos los valores en el interior de `lista_sumar`. Para eso necesitamos definir una variable que apodaremos `sum` donde vamos a ir guardando los valores de la suma, que antes de la suma tendrá valor cero (valor de inicialización). En esta parte vamos a introducir otra sintaxis: Si queremos sumar el valor actual de `sum` con otro número (el siguiente número de la lista) y guardar el valor de la suma en la misma variable, en vez de escribir `sum = sum + el`, basta con escribir `sum += el` y ya el código sabrá que debe reescribir la variable sum con el nuevo resultado de esta suma.

````python
sum = 0
for el in lista_sumar:
    sum += el
print(sum)
````

### 4.2 While

Supongamos que ahora no queremos sumar los valores de una lista sino que queremos sumar un valor continuamente a `sum` hasta que se pase cierto valor. En ese caso ya no utilizaríamos `for` porque no tenemos un grupo de elementos que recorrer, necesitaríamos usar otro tipo de lógica que nos permita hacer algo continuamente mientras se siga cumpliendo una condición. Aquí es donde entra `while` el cual cumple la siguiente lógica

```
while (condición que verificamos que se siga cumpliendo):
    haz algo continuamente
```
Así, sumemos a la variable `sum` - que habíamos definido previamente - el valor de 1 hasta que `sum = 10`, es decir, queremos que le siga sumando valores de 1 mientras que se cumpla la condición de que `sum < 10`.


````python
while (sum < 10):
    sum += 1
    print(sum)

print("Valor final de sum:", sum)
````

## 5. Funciones y funciones lambda

Hasta el momento, cada vez que hemos reutilizar la misma lógica de código, siempre hemos copiado y pegado el código y solo cambiado el nombre de la variable, como fue en el caso del control de flujo con los `if-else`. Sin embargo, eso conlleva a un gran problema; imagina la siguiente situación: estás desarrollando un código de llama esa misma lógica unas diez veces al menos, pero te das cuenta, después de ya haber copiado el código y cambiado el nombre de las variables, que tenías un "bug" (nombre técnico para un error de código) en tu código.....ahora te toca arreglarlo en cada una de las diez partes que tienes ese mismo error. Eso significa una gran inversión de tiempo que hubieras podido ahorrarte si tan solo hubiera una manera de llamar esa misma lógica y lo único que debo hacer es pasarle la variable que quiero que procese en cada caso....si tan solo existiera...¡Claro que existe! Esas son las funciones y vamos a explicarlas más a fondo en esta sección.

### 5.1 Funciones comunes

Una función "común" tiene la siguiente lógica

```
def mi_funcion(recibe unos argumentos a1, a2, etc.):
    hace algo con a1, a2, a3
    (Opcional) retorna algún valor
```

y luego simplemente llamas la función dándole los argumentos que pedí con la lógica

```
mi_funcion(valor_1, valor_2, valor_3)
```

Ilustrémoslo con una función suma que sume un par de números y nos retorne el valor de la suma.

````python
def sumar(a, b):
    return a + b
````

````python
mi_suma = sumar(3, 5)
mi_suma
````

También en vez de regresar el valor de la variable, si lo único que queremos es el valor de la suma, podemos omitir el `return` y colocar directamente el `print()`

````python
def sumar(a, b):
    print(a + b)
````

````python
sumar(3, 5)
````

En este caso dan el mismo resultado, pero recomendaría utilizar el `return` siempre que se esté interesado en utilizar el resultado de la función en otro procedimiento.

Ahora usemos como ejemplo el caso del agua a temperatura normal, congelada o ebullendo. La idea es que solo le debemos pasar a una función `how_is_my_water` (_la escribo en inglés porque al código no le gustan las tildes y no me gusta escribir sin tildes_) y me debe decir en cada ocasión cómo está mi agua. 

````python
def how_is_my_water(measurement):
    if measurement >= 100:
        return "ebullendo!"
    elif measurement < 0:
        return "congelado 🥶"
    else:
        return "temperatura normal."
````

````python
print(how_is_my_water(ocasion_1))
print(how_is_my_water(ocasion_2))
print(how_is_my_water(ocasion_3))
````

Como vieron, retornamos el estado del agua en nuestra función en cada una de las condiciones con el comando `return`. Como ese comando solo se llama cuando se cumple una de las tres condiciones, no hay problema con que lo hayamos escrito varias veces dentro de la función. Aquí también podríamos omitir el `return` y poner directamente `print()` en cada condición; como siempre, todo depende lo que les interese hacer con su función.

### 5.2 Funciones lambda

Esta es otra manera de escribir funciones en una sola línea y tiene la siguiente lógica

```
mi_funcion = lambda a1, a2, a3, ... : haz algo con a1, a2, a3, ...
```

En este caso, `mi_funcion` se creó como si fuese una variable y la clave que define la función es el comando `lambda`. Así, es como si `def mi_funcion(a1,a2,a3,...):` fuese el equivalente a `mi_funcion = lambda a1, a2, a3, ...:`. Veamos el ejemplo con la función de suma.

````python
sumar_con_lambda = lambda a, b: a + b
````

````python
mi_suma = sumar_con_lambda(3, 5)
mi_suma
````

## 6. Importar librerías

Aunque en teoría podríamos crear todas las funciones que queramos implementando todos los métodos matemáticos que conozcamos, lo más probable es que no sepamos utilizar el algoritmo matemático más óptimo para hacerlo. Por ejemplo, aunque en principio podríamos pensar que para sumar todos los números desde cero hasta un millón  deberíamos usar un loop, y en principio eso daría solución a esta tarea, el problema es cuando comencemos a subir de órdenes de magnitud y ahora queramos hacer la suma hasta cien mil millones o hasta miles de millones; ahí es cuando tenemos el problema del tiempo de cómputo del tipo de algoritmo implementado, llamado **complejidad temporal**. Este concepto es de gran importancia en un área de las matemáticas denominada **Optimización** y en resumidas cuentas nos habla de cómo resolver el mismo problema pero con un algoritmo que, parece más difícil y largo de escribir, pero que toma la mitad del tiempo de cálculo. 

Después de todo este cuento, lo que quiero concluir es que probablemente ni tú ni yo sepamos la manera más óptima de implementar cada algoritmo que queramos, así que lo mejor es llamar librerías que ya lo hayan hecho por nosotros. Una librería es básicamente un conjunto de funciones y otros objetos desarrollados por un grupo de Científicos de datos o Ingenieros de Software que utilizan los algoritmos más óptimos para resolver una ser de problemas específicos.

Para importar una librería la lógica a utilizar es

```
import librería_de_interés
```

Como ejemplo tomremos una de las librerías más usados en ciencia de datos y en python en general, llamada `numpy` que es una libería que permite hacer operaciones matemáticas rápidamente sobre vectores, matrices o tensores de datos.

````python
import numpy
````

Para usar las funciones de la librería, simplemente basta con escribir `numpy.función_que_quiero_usar(a1, a2, a3)`. Haremos esto con una multiplicación

````python
mi_mult = numpy.multiply(2, 3)
mi_mult
````

Algo interesante que pueden notar es que numpy representa los números mediante un data type propio que aunque también hace referencia a los enteros, está programado para ser más eficiente con los procesos matemáticos internos de su librería. Si quieren imprimir únicamente el número, basta con usar `print()`.

````python
print(mi_mult)
````

Otra cosa interesante de las librerías es que podemos ponerles apodos o "aliases" (como si dice comúnmente en programación). Así es más corto llamar a una librería cuando se usan constantemente funciones internas de la misma. Para apodar una librería basta con la siguiente lógica

```
import librería_de_interés as su_apodo
```
y ahora llamaras a las funciones de la librería como

`su_apodo.función_que_quiero_usar(a1, a2, a3)`

Veamos el ejemplo con `numpy` nuevamente, la cual ya está estandarizado apodarla `np`

````python
import numpy as np
````

````python
np.sum([1, 2, 3, 4, 5])
````

Aquí numpy recibió una lista y automáticamente la convirtió en un arreglo de numpy para poder realizar la suma de todos sus elementos; y como ven, esta suma la hizo sin la necesidad de que tuviéramos que hacer un for loop ni nada parecido. Si quieren saber más sobre Numpy, recomiendo visitar su [documentación oficial](https://numpy.org/doc/stable/).
